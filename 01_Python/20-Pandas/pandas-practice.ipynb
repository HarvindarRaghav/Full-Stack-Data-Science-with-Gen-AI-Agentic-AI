# 1. Series

A Series is a **one-dimensional labeled array**.

It is similar to a single column in an Excel sheet.

### Syntax

```python
pd.Series(data)


````python

import pandas as pd


marks = pd.Series([80, 75, 90, 65])

print(marks)
````

# 2. Series with Custom Index

We can provide our own index labels while creating a Series.

````python
marks = pd.Series(
    [80, 75, 90],
    index=["Rahul", "Amit", "Raj"]
)

print(marks)
````

### Access Series Value

````python
print(marks["Rahul"])
````

# 3. DataFrame

A DataFrame is a **two-dimensional table** containing rows and columns.

It is similar to an Excel spreadsheet or a database table.

### Example

| Name | Age | Salary |
|------|-----|--------|
| Rahul | 25 | 50000 |
| Amit | 30 | 60000 |
| Raj | 28 | 55000 |

````python
data = {
    "Name": ["Rahul", "Amit", "Raj"],
    "Age": [25, 30, 28],
    "Salary": [50000, 60000, 55000]
}

df = pd.DataFrame(data)

print(df)
````

## Understanding `pd.DataFrame(data)`

Here:

pd → Pandas library

df → Variable that stores the DataFrame

# 4. head() and tail()

### head()

`head()` displays the first 5 rows by default.

```python
df.head()

````python
df.head(2)
````

### tail() displays the last 5 rows by default.

````python
df.tail()
````

### We can also specify the number:

````python
df.tail(2)
````

### Difference between head() and tail(): The head() method returns the first n rows of the DataFrame, while the tail() method returns the last n rows of the DataFrame. By default, both methods return 5 rows if no argument is provided.

````python




print(df.head())
print("Now we can specify the number of rows to display using the head() method:")
print(df.head(2))
print(" We can also specify the number of rows to display using the tail() method:")
print(df.tail())
print("Difference between head() and tail(): The head() method returns the first n rows of the DataFrame, while the tail() method returns the last n rows of the DataFrame. By default, both methods return 5 rows if no argument is provided.")
print(df.tail(2))
````

# 12. Filtering Data

Filtering means selecting rows based on a condition.

### Example

Find employees whose salary is greater than 55000.

```python
df[df["Salary"] > 55000]

````python


result = df[df["Salary"] > 55000]

print(result)
````

# 13. Multiple Conditions

We can combine multiple conditions.

### AND

Use:

```python
&

### Salary greater than 55000 AND Age greater than 25:

````python
df[
    (df["Salary"] > 55000) &
    (df["Age"] > 25)
]
````

````python


result = df[
    (df["Salary"] > 55000) &
    (df["Age"] > 25)
]

print(result)
````
