# 3. DataFrame

A DataFrame is a **two-dimensional table** containing rows and columns.

It is similar to an Excel spreadsheet or a database table.

### Example

| Name | Age | Salary |
|------|-----|--------|
| Rahul | 25 | 50000 |
| Amit | 30 | 60000 |
| Raj | 28 | 55000 |

````python
data = {
    "Name": ["Rahul", "Amit", "Raj"],
    "Age": [25, 30, 28],
    "Salary": [50000, 60000, 55000]
}

df = pd.DataFrame(data)

print(df)
````

````python
data = {
    "Name": ["Rahul", "Amit", "Raj"],
    "Salary": [50000, None, 60000]
}

df_missing = pd.DataFrame(data)

print(df_missing)
````

### here isnull() method is used to check for missing values in the DataFrame. It returns a DataFrame of the same shape as df_missing, where each element is True if the corresponding value is missing (NaN) and False otherwise. The sum() method is then applied to count the number of missing values in each column.


````python
### here isnull() method is used to check for missing values in the DataFrame. It returns a DataFrame of the same shape as df_missing, where each element is True if the corresponding value is missing (NaN) and False otherwise. The sum() method is then applied to count the number of missing values in each column.
print(df_missing.isnull())
### here sum method is used to count the number of missing values in each column
print(df_missing.isnull().sum())
````

## `isnull()` Quick Revision

| Method / Value | Meaning |
|---|---|
| `isnull()` | Finds missing values |
| `True` | Missing value (`NaN`) |
| `False` | Value is present / not missing |
| `sum()` | Counts the `True` values |
| `isnull().sum()` | Gives the number of missing values in each column |

### Remember

`isnull()` → **Find missing values**

`True` → **Missing**

`False` → **Not missing**

`sum()` → **Count True values**

`isnull().sum()` → **Count missing values column-wise**

````python
df_missing["Salary"] = df_missing["Salary"].fillna(0)

print(df_missing)
````
