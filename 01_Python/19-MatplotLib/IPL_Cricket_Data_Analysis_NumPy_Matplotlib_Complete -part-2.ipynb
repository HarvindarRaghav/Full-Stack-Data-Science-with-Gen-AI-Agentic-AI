# 1️⃣ Import Libraries

````python
import numpy as np
import matplotlib.pyplot as plt
````

# 2️⃣ Seasons and Players

`Seasons` is used as the X-axis.

`Sdict` maps season names to column indexes.

`Players` contains player names.

`Pdict` maps player names to row indexes.

**Key idea:** Row = Player, Column = Season.


````python
Seasons = ["2015","2016","2017","2018","2019","2020","2021","2022","2023","2024"]

Sdict = {"2015":0,"2016":1,"2017":2,"2018":3,"2019":4,
         "2020":5,"2021":6,"2022":7,"2023":8,"2024":9}

Players = ["Sachin","Rahul","Smith","Sami","Pollard","Morris","Samson","Dhoni","Kohli","Sky"]

Pdict = {"Sachin":0,"Rahul":1,"Smith":2,"Sami":3,"Pollard":4,
         "Morris":5,"Samson":6,"Dhoni":7,"Kohli":8,"Sky":9}
````

# 3️⃣ Salary Data

`Salary[player_index, season_index]` selects one value.


````python
Sachin_Salary = [15946875,17718750,19490625,21262500,23034375,24806250,25244493,27849149,30453805,23500000]
Rahul_Salary = [12000000,12744189,13488377,14232567,14976754,16324500,18038573,19752645,21466718,23180790]
Smith_Salary = [4621800,5828090,13041250,14410581,15779912,14500000,16022500,17545000,19067500,20644400]
Sami_Salary = [3713640,4694041,13041250,14410581,15779912,17149243,18518574,19450000,22407474,22458000]
Pollard_Salary = [4493160,4806720,6061274,13758000,15202590,16647180,18091770,19536360,20513178,21436271]
Morris_Salary = [3348000,4235220,12455000,14410581,15779912,14500000,16022500,17545000,19067500,20644400]
Samson_Salary = [3144240,3380160,3615960,4574189,13520500,14940153,16359805,17779458,18668431,20068563]
Dhoni_Salary = [0,0,4171200,4484040,4796880,6053663,15506632,16669630,17832627,18995624]
Kohli_Salary = [0,0,0,4822800,5184480,5546160,6993708,16402500,17632688,18862875]
Sky_Salary = [3031920,3841443,13041250,14410581,15779912,14200000,15691000,17182000,18673000,15000000]

Salary = np.array([Sachin_Salary,Rahul_Salary,Smith_Salary,Sami_Salary,Pollard_Salary,
                   Morris_Salary,Samson_Salary,Dhoni_Salary,Kohli_Salary,Sky_Salary])

print("Salary shape:", Salary.shape)
````

# 4️⃣ Games and Points Data

The `Games` and `Points` matrices use the same structure:

- Rows → players
- Columns → seasons


````python
Games = np.array([
    [80,77,82,82,73,82,58,78,6,35],
    [82,57,82,79,76,72,60,72,79,80],
    [79,78,75,81,76,79,62,76,77,69],
    [80,65,77,66,69,77,55,67,77,40],
    [82,82,82,79,82,78,54,76,71,41],
    [70,69,67,77,70,77,57,74,79,44],
    [78,64,80,78,45,80,60,70,62,82],
    [35,35,80,74,82,78,66,81,81,27],
    [40,40,40,81,78,81,39,0,10,51],
    [75,51,51,79,77,76,49,69,54,62]
])

Points = np.array([
    [2832,2430,2323,2201,1970,2078,1616,2133,83,782],
    [1653,1426,1779,1688,1619,1312,1129,1170,1245,1154],
    [2478,2132,2250,2304,2258,2111,1683,2036,2089,1743],
    [2122,1881,1978,1504,1943,1970,1245,1920,2112,966],
    [1292,1443,1695,1624,1503,1784,1113,1296,1297,646],
    [1572,1561,1496,1746,1678,1438,1025,1232,1281,928],
    [1258,1104,1684,1781,841,1268,1189,1186,1185,1564],
    [903,903,1624,1871,2472,2161,1850,2280,2593,686],
    [597,597,597,1361,1619,2026,852,0,159,904],
    [2040,1397,1254,2386,2045,1941,1082,1463,1028,1331]
])

print("Games shape:", Games.shape)
print("Points shape:", Points.shape)
````

# 5️⃣ NumPy Indexing & Slicing

Examples:

- `Salary[0]` → Sachin's salary history
- `Salary[Pdict["Rahul"]]` → Rahul's salary history
- `Salary[:, 0]` → all players in 2015
- `Salary[1, 5]` → Rahul's 2020 salary
- `Salary[:, 2:5]` → all players from 2017–2019


````python
print("Sachin:", Salary[Pdict["Sachin"]])
print("Rahul:", Salary[Pdict["Rahul"]])
print("All players in 2015:", Salary[:, Sdict["2015"]])
print("Rahul salary in 2020:", Salary[Pdict["Rahul"], Sdict["2020"]])
````

# 6️⃣ `axis=0` vs `axis=1`

`axis=0` → operate down rows → one result per column/season.

`axis=1` → operate across columns → one result per row/player.

### 🧠 Memory

```text
axis=0 → Season-wise
axis=1 → Player-wise
```


````python
season_total_points = np.sum(Points, axis=0)
player_total_points = np.sum(Points, axis=1)

print("Season totals:", season_total_points)
print("Player totals:", player_total_points)
````

# 7️⃣ Player Ranking

`np.argsort()` returns indexes that sort an array.

`[::-1]` reverses the order so the highest values appear first.


````python
ranking = np.argsort(player_total_points)[::-1]

for index in ranking:
    print(Players[index], player_total_points[index])
````

# 8️⃣ Performance Metrics

### Points per Game

`Total Points / Total Games`

### Salary per Game

`Total Salary / Total Games`

These metrics help compare efficiency, not only totals.


````python
total_games = np.sum(Games, axis=1)
total_salary = np.sum(Salary, axis=1)

points_per_game = np.divide(
    player_total_points, total_games,
    out=np.zeros_like(player_total_points, dtype=float),
    where=total_games != 0
)

salary_per_game = np.divide(
    total_salary, total_games,
    out=np.zeros_like(total_salary, dtype=float),
    where=total_games != 0
)

for i, player in enumerate(Players):
    print(f"{player}: {points_per_game[i]:.2f} points/game | {salary_per_game[i]:.2f} salary/game")
````

# 9️⃣ Matplotlib `plot()` — Line Chart

General pattern:

```text
plt.plot(X, Y, marker, color, linestyle, label)
```

For this project:
- X → `Seasons`
- Y → player salary/points/games
- `marker` → point shape
- `c` → color
- `ls` → line style
- `label` → line name
- `plt.legend()` → displays labels


````python
plt.figure(figsize=(12, 6))

plt.plot(
    Seasons,
    Salary[Pdict["Rahul"]],
    label="Rahul",
    marker="o",
    ls="-.",
    c="red",
    lw=2,
    ms=7
)

plt.xlabel("Season")
plt.ylabel("Salary")
plt.title("Rahul Salary by Season")
plt.grid()
plt.legend()
plt.tight_layout()
plt.show()
````

# 1️⃣1️⃣ Bar Chart — Total Points

A bar chart is useful for comparing categories.


````python
total_points = np.sum(Points, axis=1)

plt.figure(figsize=(12, 6))
plt.bar(Players, total_points)
plt.xlabel("Players")
plt.ylabel("Total Points")
plt.title("Total Points by Player")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

## 🔍 Observation

Use this chart to identify the highest and lowest total performers and compare the size of the differences.

**Data Science Insight:** Bar charts answer category-comparison questions.


# 1️⃣2️⃣ Horizontal Bar Chart — Ranking

`plt.barh()` is useful for rankings because category names are easier to read.


````python
ranking = np.argsort(total_points)

plt.figure(figsize=(10, 6))
plt.barh(np.array(Players)[ranking], total_points[ranking])
plt.xlabel("Total Points")
plt.ylabel("Players")
plt.title("IPL Player Ranking by Total Points")
plt.tight_layout()
plt.show()
````

## 🔍 Observation

The ranking chart makes top and bottom performers easy to identify.

**Data Science Insight:** Ranking visualizations are useful for communicating decisions quickly.


# 1️⃣3️⃣ Line Chart — Total Points by Season

A line chart is ideal for trends over time.


````python
season_total_points = np.sum(Points, axis=0)

plt.figure(figsize=(12, 6))
plt.plot(Seasons, season_total_points, marker="o", label="Total Points")
plt.xlabel("Season")
plt.ylabel("Total Points")
plt.title("Total IPL Points by Season")
plt.grid()
plt.legend()
plt.tight_layout()
plt.show()
````

## 🔍 Observation

Look for peaks, drops and overall direction.

**Data Science Insight:** Line charts are especially useful for ordered/time-based data.


# 1️⃣4️⃣ Scatter Plot — Salary vs Points

A scatter plot investigates the relationship between two numerical variables.


````python
plt.figure(figsize=(10, 6))
plt.scatter(total_salary, total_points)

for i, player in enumerate(Players):
    plt.annotate(player, (total_salary[i], total_points[i]))

plt.xlabel("Total Salary")
plt.ylabel("Total Points")
plt.title("Total Salary vs Total Points")
plt.tight_layout()
plt.show()
````

## 🔍 Observation

Look for players with high salary/high points, high salary/low points, and low salary/high points.

**Data Science Insight:** A visual pattern is not proof of correlation; statistical analysis can measure the relationship.


# 1️⃣5️⃣ Histogram — Points Distribution

A histogram shows how numerical values are distributed.


````python
plt.figure(figsize=(10, 6))
plt.hist(total_points, bins=6)
plt.xlabel("Total Points")
plt.ylabel("Number of Players")
plt.title("Distribution of Total Player Points")
plt.tight_layout()
plt.show()
````

## 🔍 Observation

Check whether points are concentrated, spread out, or skewed.

**Data Science Insight:** Histograms answer distribution questions.


# 1️⃣6️⃣ Boxplot — Points Distribution

A boxplot summarizes median, quartiles, spread and potential outliers.


````python
plt.figure(figsize=(10, 6))
plt.boxplot(Points.T)
plt.xlabel("Players")
plt.ylabel("Points")
plt.title("Distribution of IPL Points")
plt.xticks(range(1, len(Players)+1), Players, rotation=45)
plt.tight_layout()
plt.show()
````

## 🔍 Observation

Look for players with high variation, consistency and unusual seasons.

**Data Science Insight:** Boxplots are useful when spread and outliers matter.


# 1️⃣7️⃣ Heatmap — IPL Points

A heatmap visualizes a matrix.

- Rows → Players
- Columns → Seasons
- Intensity → Points value


````python
plt.figure(figsize=(12, 7))
plt.imshow(Points)
plt.colorbar()
plt.xticks(range(len(Seasons)), Seasons)
plt.yticks(range(len(Players)), Players)
plt.xlabel("Season")
plt.ylabel("Player")
plt.title("IPL Points Heatmap")
plt.tight_layout()
plt.show()
````

## 🔍 Observation

The heatmap helps identify strong/weak seasons, consistency and large changes.

**Data Science Insight:** Heatmaps are useful when many values must be compared across two dimensions.


# 1️⃣8️⃣ Figure vs Axes

**Figure** = complete canvas.

**Axes** = actual plotting area.

Modern Matplotlib often uses:

```python
fig, ax = plt.subplots()
```

This becomes especially useful with multiple charts.


````python
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(Seasons, Salary[Pdict["Rahul"]], marker="o", label="Rahul")
ax.set_xlabel("Season")
ax.set_ylabel("Salary")
ax.set_title("Rahul Salary Trend")
ax.legend()

plt.show()
````

# 1️⃣9️⃣ Subplots

Subplots allow several charts inside one figure.

```text
┌───────────┬───────────┐
│  Chart 1  │  Chart 2  │
├───────────┼───────────┤
│  Chart 3  │  Chart 4  │
└───────────┴───────────┘
```


````python
fig, ax = plt.subplots(2, 2, figsize=(14, 10))

ax[0, 0].plot(Seasons, Salary[Pdict["Rahul"]], marker="o")
ax[0, 0].set_title("Rahul Salary")

ax[0, 1].plot(Seasons, Points[Pdict["Rahul"]], marker="o")
ax[0, 1].set_title("Rahul Points")

ax[1, 0].plot(Seasons, Games[Pdict["Rahul"]], marker="o")
ax[1, 0].set_title("Rahul Games")

rahul_ppg = np.divide(
    Points[Pdict["Rahul"]], Games[Pdict["Rahul"]],
    out=np.zeros_like(Points[Pdict["Rahul"]], dtype=float),
    where=Games[Pdict["Rahul"]] != 0
)
ax[1, 1].plot(Seasons, rahul_ppg, marker="o")
ax[1, 1].set_title("Rahul Points/Game")

plt.tight_layout()
plt.show()
````

# 2️⃣0️⃣ Annotations

`plt.annotate()` highlights important values on a graph.

Typical uses:
- Highest value
- Lowest value
- Important season
- Unusual observation


````python
rahul_salary = Salary[Pdict["Rahul"]]
max_index = np.argmax(rahul_salary)

plt.figure(figsize=(12, 6))
plt.plot(Seasons, rahul_salary, marker="o", label="Rahul")
plt.annotate(
    f"Highest: {rahul_salary[max_index]:,}",
    xy=(Seasons[max_index], rahul_salary[max_index])
)
plt.xlabel("Season")
plt.ylabel("Salary")
plt.title("Rahul Salary Trend with Highest Value")
plt.legend()
plt.tight_layout()
plt.show()
````

# 2️⃣1️⃣ Save a Visualization

`plt.savefig()` saves the chart for reports, presentations and documentation.


````python
plt.figure(figsize=(10, 6))
plt.plot(Seasons, Salary[Pdict["Rahul"]], marker="o", label="Rahul")
plt.xlabel("Season")
plt.ylabel("Salary")
plt.title("Rahul Salary Trend")
plt.legend()
plt.tight_layout()
plt.savefig("rahul_salary_trend.png")
plt.show()
````

# 2️⃣2️⃣ `diff()` and `cumsum()`

`np.diff()` calculates consecutive changes.

`np.cumsum()` calculates a running cumulative total.


````python
rahul_salary_change = np.diff(Salary[Pdict["Rahul"]])
rahul_cumulative_points = np.cumsum(Points[Pdict["Rahul"]])

print("Salary changes:", rahul_salary_change)
print("Cumulative points:", rahul_cumulative_points)
````

# 2️⃣3️⃣ Common Errors
