# Matplotlib Notes

Matplotlib is a Python library used to create charts, graphs, and visualizations. The `pyplot` module provides simple functions for creating plots in Jupyter Notebook. It is commonly used with NumPy and Pandas for data analysis and with image arrays for image visualization.

This notebook covers the most important Matplotlib methods with explanations, comments, examples, and a quick revision section.

````python
# Import Matplotlib's pyplot module
import matplotlib.pyplot as plt

# Import NumPy for creating example data
import numpy as np
````

## 1. Basic Line Plot

A line plot connects data points with lines. It is useful for showing trends, changes over time, and relationships between two numerical variables.

### Important method
- `plt.plot(x, y)` → Creates a line plot
- `plt.title()` → Adds a title
- `plt.xlabel()` → Adds an X-axis label
- `plt.ylabel()` → Adds a Y-axis label
- `plt.show()` → Displays the plot

````python
# Example: Basic line plot
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
plt.title("Basic Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
````

## 2. Figure and Figure Size

A figure is the overall drawing area that contains one or more plots. `figsize` controls the width and height of the figure in inches.

### Methods
- `plt.figure()` → Creates a new figure
- `plt.figure(figsize=(width, height))` → Sets the figure size

````python
# Create a figure with a custom size
plt.figure(figsize=(8, 5))
plt.plot(x, y)
plt.title("Custom Figure Size")
plt.show()
````

## 3. Scatter Plot

A scatter plot displays individual points instead of connecting them with lines. It is useful for finding relationships, clusters, and patterns between variables.

### Method
- `plt.scatter(x, y)` → Creates a scatter plot

````python
# Example: Scatter plot
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
````

## 4. Bar Chart

A bar chart compares values across categories. `plt.bar()` creates vertical bars, while `plt.barh()` creates horizontal bars.

### Methods
- `plt.bar(x, height)` → Vertical bar chart
- `plt.barh(y, width)` → Horizontal bar chart

````python
# Example: Vertical bar chart
names = ["Java", "Python", "AWS", "Docker"]
values = [80, 90, 70, 60]

plt.bar(names, values)
plt.title("Technology Skills")
plt.xlabel("Technology")
plt.ylabel("Score")
plt.show()
````

````python
# Example: Horizontal bar chart
plt.barh(names, values)
plt.title("Technology Skills")
plt.xlabel("Score")
plt.ylabel("Technology")
plt.show()
````

## 5. Histogram

A histogram shows the distribution of numerical data. The `bins` argument divides the values into intervals and counts how many values fall into each interval.

### Methods
- `plt.hist(data)` → Creates a histogram
- `plt.hist(data, bins=10)` → Controls the number of bins

````python
# Example: Histogram
ages = [22, 25, 25, 28, 30, 30, 32, 35, 35, 40, 42, 45]

plt.hist(ages, bins=5)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
````

## 6. Pie Chart

A pie chart represents parts of a whole. It is useful when the categories represent percentages or proportions.

### Methods
- `plt.pie(data)` → Creates a pie chart
- `labels=` → Adds category labels
- `autopct=` → Displays percentages

````python
# Example: Pie chart
labels = ["Java", "Python", "AWS", "Docker"]
values = [30, 40, 20, 10]

plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.title("Technology Distribution")
plt.show()
````

## 7. Grid and Legend

A grid helps read values from a chart. A legend identifies multiple lines or data series in the same plot.

### Methods
- `plt.grid()` → Shows grid lines
- `plt.legend()` → Shows the legend
- `label=` → Gives a data series a name

````python
# Example: Grid and legend
y1 = [10, 20, 15, 30, 25]
y2 = [8, 18, 20, 25, 30]

plt.plot(x, y1, label="Series 1")
plt.plot(x, y2, label="Series 2")
plt.title("Multiple Lines")
plt.xlabel("X Values")
plt.ylabel("Values")
plt.grid()
plt.legend()
plt.show()
````

## 8. Axis Limits and Ticks

Axis methods control the visible range and labels on the X and Y axes.

### Methods
- `plt.xlim()` → Sets X-axis limits
- `plt.ylim()` → Sets Y-axis limits
- `plt.xticks()` → Controls X-axis tick positions/labels
- `plt.yticks()` → Controls Y-axis tick positions/labels

````python
# Example: Control axis limits
plt.plot(x, y)
plt.xlim(1, 5)
plt.ylim(0, 35)
plt.xticks([1, 2, 3, 4, 5])
plt.title("Axis Control")
plt.show()
````

## 9. Text and Annotation

`plt.text()` places text at a specified coordinate. `plt.annotate()` can add explanatory text with an arrow pointing to a particular data point.

### Methods
- `plt.text(x, y, text)` → Adds text
- `plt.annotate(text, xy=...)` → Adds an annotation

````python
# Example: Add text and annotation
plt.plot(x, y, marker="o")
plt.text(3, 15, "Important Point")
plt.annotate("Highest", xy=(4, 30), xytext=(3.2, 33), arrowprops={"arrowstyle": "->"})
plt.title("Text and Annotation")
plt.show()
````

## 10. Display an Image

`plt.imshow()` displays an image or a 2D/3D NumPy array. For an RGB image, the array commonly has shape `(height, width, 3)`. For a grayscale image, `cmap="gray"` is commonly used.

### Methods / options
- `plt.imshow(image_arr)` → Displays an image
- `cmap="gray"` → Uses a grayscale colormap
- `plt.axis("off")` → Hides axes
- `plt.colorbar()` → Displays a color scale

````python
# Example image array
image_arr = np.random.randint(0, 256, (200, 300, 3), dtype=np.uint8)

# Display the image
plt.imshow(image_arr)
plt.axis("off")
plt.title("Image Display")
plt.show()
````

````python
# Display an image as grayscale
gray_image = np.random.randint(0, 256, (200, 300), dtype=np.uint8)

plt.imshow(gray_image, cmap="gray")
plt.axis("off")
plt.title("Grayscale Image")
plt.show()
````

## 11. Colorbar

A colorbar explains how numerical values map to colors in an image or heatmap. It is especially useful for 2D numerical arrays.

### Method
- `plt.colorbar()` → Adds a color scale

````python
# Example: Image with colorbar
data = np.random.randint(0, 256, (100, 100))

plt.imshow(data, cmap="gray")
plt.colorbar()
plt.title("Image with Colorbar")
plt.show()
````

## 12. Subplot

`plt.subplot()` allows multiple plots to be placed inside one figure. The basic syntax is `plt.subplot(rows, columns, position)`.

### Method
- `plt.subplot(1, 2, 1)` → First plot in a 1-row, 2-column layout
- `plt.subplot(1, 2, 2)` → Second plot in the same layout

````python
# Example: Two plots in one figure
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(x, y)
plt.title("Line Plot")

plt.subplot(1, 2, 2)
plt.bar(names, values)
plt.title("Bar Chart")

plt.tight_layout()
plt.show()
````

## 13. Subplots API

`plt.subplots()` is a more flexible and commonly preferred approach for working with multiple axes. It returns a figure object and one or more axes objects.

### Methods
- `plt.subplots()` → Creates a figure and axes
- `ax.plot()` → Plot on a specific axes
- `ax.imshow()` → Display an image on a specific axes
- `ax.set_title()` → Set axes title
- `ax.set_xlabel()` → Set X label
- `ax.set_ylabel()` → Set Y label

````python
# Example: Object-oriented subplots
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(x, y)
axes[0].set_title("Line Plot")

axes[1].bar(names, values)
axes[1].set_title("Bar Chart")

plt.tight_layout()
plt.show()
````

## 14. Save a Plot

`plt.savefig()` saves the current figure to a file. The `dpi` parameter controls the resolution for raster image formats.

### Methods
- `plt.savefig("output.png")` → Save as PNG
- `plt.savefig("output.jpg")` → Save as JPG
- `plt.savefig("output.pdf")` → Save as PDF
- `dpi=300` → Save with higher resolution

````python
# Example: Save a plot
plt.plot(x, y)
plt.title("Plot to Save")
plt.savefig("matplotlib_output.png", dpi=300, bbox_inches="tight")
plt.show()
````

## 15. Common Plot Customization

Matplotlib allows you to customize markers, line styles, transparency, and labels. These options make charts easier to understand.

### Common parameters
- `marker` → Changes point marker
- `linestyle` → Changes line style
- `linewidth` → Changes line width
- `alpha` → Controls transparency
- `label` → Gives a series a name

````python
# Example: Common customization parameters
plt.plot(
    x, y,
    marker="o",
    linestyle="--",
    linewidth=2,
    alpha=0.8,
    label="Sales"
)
plt.title("Customized Line Plot")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid()
plt.show()
````

- You want to show trends
- You want to compare changes over time

```python

plt.xlabel("Month")
plt.ylabel("Sales")
