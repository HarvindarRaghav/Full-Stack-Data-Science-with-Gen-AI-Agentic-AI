## 12. Total Salary Per Player

`axis=1` performs the operation across columns, producing one result per player.

````python
total_salary = np.sum(Salary, axis=1)

for i in range(len(Players)):
    print(Players[i], total_salary[i])
````

## 17. Minimum and Maximum

`np.min()` and `np.max()` find minimum and maximum values.

````python
print("Highest salary overall:", np.max(Salary))
print("Lowest salary overall:", np.min(Salary))
print("Highest points overall:", np.max(Points))
print("Lowest points overall:", np.min(Points))

print("Maximum salary per player:")
print(np.max(Salary, axis=1))

print("Maximum points per player:")
print(np.max(Points, axis=1))
````

## 18. Highest Scoring Player

`np.argmax()` returns the index of the maximum value.

````python
total_points = np.sum(Points, axis=1)
index = np.argmax(total_points)

print("Highest Scoring Player:", Players[index])
print("Total Points:", total_points[index])
````

## 19. Highest Paid Player

````python
total_salary = np.sum(Salary, axis=1)
index = np.argmax(total_salary)

print("Highest Paid Player:", Players[index])
print("Total Salary:", total_salary[index])
````

## 20. Player Ranking Using `argsort()`

`np.argsort()` returns indexes that sort an array. `[::-1]` changes ascending order to descending order.

````python
total_points = np.sum(Points, axis=1)
ranking = np.argsort(total_points)[::-1]

for rank, index in enumerate(ranking, start=1):
    print(rank, Players[index], total_points[index])
````

## 21. Top 3 Players

````python
for index in ranking[:3]:
    print(Players[index], total_points[index])
````

## 22. Boolean Indexing

A condition such as `Points > 2000` creates a Boolean array. It can be used to filter values.

````python
result = Points[Points > 2000]
print(result)
````

## 23. Count Values Greater Than 2000

````python
count = np.sum(Points > 2000)
print("Number of point values greater than 2000:", count)
````

## 24. Find Players With Total Points Above a Threshold

````python
total_points = np.sum(Points, axis=1)

condition = total_points > 15000
selected_players = np.array(Players)[condition]

print(selected_players)
````

## 25. `np.where()`

`np.where()` can return indexes or select values based on a condition.

````python
indexes = np.where(total_points > 15000)[0]

for index in indexes:
    print(Players[index], total_points[index])
````

## 26. Points Per Game

Formula: `Points / Games`. `np.divide()` safely handles zero-game entries.

````python
points_per_game = np.divide(
    Points,
    Games,
    out=np.zeros_like(Points, dtype=float),
    where=Games != 0
)

print(points_per_game)
````

## 27. Salary Per Game

````python
salary_per_game = np.divide(
    Salary,
    Games,
    out=np.zeros_like(Salary, dtype=float),
    where=Games != 0
)

print(salary_per_game)
````

## 28. Salary Per Point

````python
salary_per_point = np.divide(
    Salary,
    Points,
    out=np.zeros_like(Salary, dtype=float),
    where=Points != 0
)

print(salary_per_point)
````

## 29. Player Efficiency

Here efficiency is defined as total points divided by total games.

````python
total_points = np.sum(Points, axis=1)
total_games = np.sum(Games, axis=1)

efficiency = np.divide(
    total_points,
    total_games,
    out=np.zeros_like(total_points, dtype=float),
    where=total_games != 0
)

for i in np.argsort(efficiency)[::-1]:
    print(Players[i], efficiency[i])
````

## 30. Year-over-Year Growth Using `np.diff()`

`np.diff()` calculates the difference between consecutive seasons.

````python
kohli_salary = Salary[Pdict["Kohli"]]

growth = np.diff(kohli_salary)

print("Kohli salary growth:")
print(growth)
````

## 31. Percentage Growth

````python
kohli_salary = Salary[Pdict["Kohli"]]

previous = kohli_salary[:-1]
growth_percent = np.divide(
    np.diff(kohli_salary) * 100,
    previous,
    out=np.zeros_like(previous, dtype=float),
    where=previous != 0
)

print(growth_percent)
````

## 32. Cumulative Points Using `cumsum()`

````python
kohli_points = Points[Pdict["Kohli"]]

cumulative_points = np.cumsum(kohli_points)

for season, value in zip(Seasons, cumulative_points):
    print(season, value)
````

## 33. Statistical Analysis

Use `mean`, `median`, `std`, and `var` for statistical analysis.

## 34. Find Position of Maximum Value

`np.unravel_index()` converts the flattened maximum position into `(row, column)`.

````python
index = np.unravel_index(np.argmax(Salary), Salary.shape)

player_index = index[0]
season_index = index[1]

print("Player:", Players[player_index])
print("Season:", Seasons[season_index])
print("Salary:", Salary[index])
````

## 35. Broadcasting

NumPy broadcasting allows operations between arrays with compatible shapes. Here we divide each player's points by their total games.

````python
total_points = np.sum(Points, axis=1)
total_games = np.sum(Games, axis=1)

points_per_game_total = total_points / total_games

for i in range(len(Players)):
    print(Players[i], points_per_game_total[i])
````

## 36. Replace Zero Values

A zero can represent no participation. We can replace zeros with `np.nan` for analysis where appropriate.

````python
salary_float = Salary.astype(float)
salary_float[salary_float == 0] = np.nan

print(salary_float)
print("Mean salary ignoring NaN:", np.nanmean(salary_float))
````

# 🔑 Most Important NumPy Concept — Axis

`axis` is one of the most important concepts in NumPy interviews.

Suppose:

```text
Points.shape = (10, 10)
```

There are:

- 10 rows → players
- 10 columns → seasons

### axis=0

Operation happens **down the rows**.

Result → one value for every column/season.

```python
np.sum(Points, axis=0)
```

Answer:

```text
Total points for each season
```

### axis=1

Operation happens **across the columns**.

Result → one value for every row/player.

```python
np.sum(Points, axis=1)
```

Answer:

```text
Total points for each player
```

### Easy memory trick

```text
axis=0 → result follows columns
axis=1 → result follows rows
```

Or remember:

```text
axis=0 → Season-wise
axis=1 → Player-wise
```

```python
np.argmax(arr)
```

returns the **index** of the maximum value.

### Q4. Difference between `argsort()` and `sort()`?

`sort()` sorts values.

`argsort()` returns the indexes that produce the sorted order.

### Q5. Why use Boolean indexing?

It allows conditional filtering:

```python
Points[Points > 2000]
```

### Q6. Why use `np.divide()` instead of `/`?

`np.divide()` can be configured to avoid division-by-zero problems.

### Q7. What is broadcasting?

Broadcasting allows NumPy to perform operations on arrays with compatible shapes without manually expanding the arrays.

### Q8. What does `flatten()` do?

It converts a multidimensional array into a 1D array.

```python
Points.flatten()
```
