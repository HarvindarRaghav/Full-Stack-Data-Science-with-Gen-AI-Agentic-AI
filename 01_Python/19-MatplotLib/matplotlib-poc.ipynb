# ⭐ Matplotlib Markers, Colors & Line Styles — Quick Revision

## 🔹 1. Markers

Markers define the **shape of each data point** on a graph.

| Marker | Name | Usage |
|---|---|---|
| `o` | Circle | `marker='o'` |
| `.` | Point | `marker='.'` |
| `,` | Pixel | `marker=','` |
| `x` | X | `marker='x'` |
| `+` | Plus | `marker='+'` |
| `*` | Star | `marker='*'` |
| `s` | Square | `marker='s'` |
| `D` | Diamond | `marker='D'` |
| `d` | Thin Diamond | `marker='d'` |
| `^` | Triangle Up | `marker='^'` |
| `v` | Triangle Down | `marker='v'` |
| `<` | Triangle Left | `marker='<'` |
| `>` | Triangle Right | `marker='>'` |
| `p` | Pentagon | `marker='p'` |
| `h` | Hexagon | `marker='h'` |
| `H` | Hexagon 2 | `marker='H'` |
| `8` | Octagon | `marker='8'` |

# 🎯 IPL Project Pattern

The general syntax is:

`plt.plot(X, Y, marker, color, linestyle, label)`

For our IPL project:

- `X` → Seasons
- `Y` → Salary / Games / Points
- `marker` → Shape of each season's point
- `color` → Line color
- `linestyle` → Connecting line style
- `label` → Player name

`plt.legend()` is used to display the labels on the graph.

### Python Code Cell Example here 

```python
import matplotlib.pyplot as plt

plt.plot(
    Seasons,
    Salary[Pdict["Rahul"]],
    label="Rahul",
    marker="o",
    ls="-.",
    c="red",
    lw=2,
    ms=8
)

plt.xlabel("Season")
plt.ylabel("Salary")
plt.title("Rahul Salary by Season")

plt.legend()
plt.show()
```

### 🐍 Another Python Code Cell — Compare Players

```python
#  c, ls ,label, ms are used to change the color, line style, label and marker size of the graph respectively
plt.plot(
    Seasons,
    Salary[Pdict["Sachin"]],
    marker="o",
    c="blue",
    ls="-",
    label="Sachin",
    ms =4
)

plt.plot(
    Seasons,
    Salary[Pdict["Rahul"]],
    marker="s",
    c="red",
    ls="--",
    label="Rahul"
)

plt.xlabel("Season")
plt.ylabel("Salary")
plt.title("Sachin vs Rahul Salary")

plt.legend()
plt.show()
```
