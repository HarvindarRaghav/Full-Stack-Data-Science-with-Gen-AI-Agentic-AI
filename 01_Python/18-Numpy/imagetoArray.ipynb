1. Display Image

## Display Image

`plt.imshow()` displays an image using Matplotlib.

`plt.show()` renders the image/output in the notebook.


```python
import matplotlib.pyplot as plt

plt.imshow(image_arr)
plt.show()
```

2. Convert Image to NumPy Array

## Convert Image to NumPy Array

`np.asarray()` converts a PIL image into a NumPy array.

The array contains pixel values of the image.


```python
import numpy as np

image_arr = np.asarray(image)

image_arr
```

3. Check Image Shape

## Image Shape

`shape` tells us the dimensions of the image.

For a color image:

`(Height, Width, Channels)`

Usually, RGB image channels = 3.


```python
print(image_arr.shape)
```

4. Check Number of Dimensions

## Number of Dimensions

`ndim` returns the number of dimensions in the NumPy array.

RGB image → 3 dimensions

Grayscale image → 2 dimensions


```python
print(image_arr.ndim)
```

5. Check Data Type

## Image Data Type

`dtype` tells us the data type used to store pixel values.

Most images use `uint8`, where pixel values range from 0 to 255.


```python
print(image_arr.dtype)
```

6. Get Image Height and Width

## Image Height and Width

The first value represents height and the second represents width.

For RGB images, the third value represents color channels.


```python
height, width, channels = image_arr.shape

print("Height:", height)
print("Width:", width)
print("Channels:", channels)
```

7. Display Image Without Axis

## Hide Axis

`plt.axis("off")` removes the x-axis and y-axis from the image display.


```python
plt.imshow(image_arr)
plt.axis("off")
plt.show()
```

8. Change Image Size

## Change Figure Size

`figsize` controls the size of the displayed image.

The values are specified as `(width, height)` in inches.


```python
plt.figure(figsize=(8, 6))
plt.imshow(image_arr)
plt.axis("off")
plt.show()
```

9. Display Grayscale Image

## Display Grayscale Image

`cmap="gray"` displays the image using a grayscale color map.

This is useful for black-and-white images.


```python
plt.imshow(image_arr, cmap="gray")
plt.axis("off")
plt.show()
```

#10.  Get Pixel Value

## Access Pixel Value

A NumPy image array can be accessed using row and column indexes.

For an RGB image:

`image_arr[row, column]`

returns the RGB values of that pixel.


```python
pixel = image_arr[100, 200]

print(pixel)
```
